```python
import pickle

df = df.reset_index(drop = True)
```

```python
indices = pd.Series(df.index,index = df['title']).drop_duplicates()
indices
```

```python
load_model = pickle.load(open('embeddings.pkl', 'rb'))
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def normalize_title(value):
    return ''.join(character for character in str(value).casefold() if character.isalnum())


def recommend(title, n=10):
    title_key = normalize_title(title)
    title_keys = df['title'].map(normalize_title)
    matches = df.index[title_keys == title_key]

    if len(matches) == 0:
        return ['Movie not found']

    idx = matches[0]
    sim_scores = cosine_similarity(load_model[idx].reshape(1, -1), load_model).flatten()
    sim_idx = sim_scores.argsort()[::-1][1:n + 1]
    return df['title'].iloc[sim_idx].tolist()
```

```python
recommend('',10)
```

```python
recommend('spiderman',10)
```
